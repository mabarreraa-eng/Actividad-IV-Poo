```python

import ipywidgets as widgets
from IPython.display import display, clear_output


# Clase Vendedor
class Vendedor:

    def __init__(self, nombre, apellidos):
        self.nombre = nombre
        self.apellidos = apellidos

    # Metodo para verificar la edad
    def verificar_edad(self, edad):

        if edad < 18:
            raise ValueError("El vendedor debe ser mayor de 18 años")

        if edad >= 0 and edad < 120:
            return edad
        else:
            raise ValueError("La edad no puede ser negativa ni mayor a 120")

    # Metodo para mostrar los datos
    def imprimir(self, edad):
        print("DATOS DEL VENDEDOR")
        print("Nombre:", self.nombre)
        print("Apellidos:", self.apellidos)
        print("Edad:", edad)


# Titulo de la interfaz
titulo = widgets.HTML(
    value="""
    <h2 style="color:#FFFFFF; text-align:center;">
        Registro de vendedor
    </h2>
    <p style="color:#FFFFFF; text-align:center;">
        Ejercicio 6.5. Lanzamiento de excepciones
    </p>
    """
)


# Campos para ingresar los datos
nombre = widgets.Text(
    description="Nombre:",
    placeholder="Ingrese el nombre",
    layout=widgets.Layout(width="90%")
)

apellidos = widgets.Text(
    description="Apellidos:",
    placeholder="Ingrese los apellidos",
    layout=widgets.Layout(width="90%")
)

edad = widgets.Text(
    description="Edad:",
    placeholder="Ingrese la edad",
    layout=widgets.Layout(width="90%")
)


# Espacio para mostrar resultados
salida = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #75B798",
        padding="15px",
        width="95%"
    )
)


# Funcion para registrar el vendedor
def registrar(b):

    with salida:
        clear_output()

        try:
            n = nombre.value.strip()
            a = apellidos.value.strip()

            if n == "" or a == "":
                raise ValueError("Debe ingresar el nombre y los apellidos")

            e = int(edad.value)

            vendedor = Vendedor(n, a)

            e = vendedor.verificar_edad(e)

            print("Vendedor registrado correctamente\n")
            vendedor.imprimir(e)

        except ValueError as error:
            print("No se pudo registrar el vendedor")
            print("Error:", error)


# Funcion para limpiar
def limpiar(b):

    nombre.value = ""
    apellidos.value = ""
    edad.value = ""

    with salida:
        clear_output()


# Boton registrar
boton_registrar = widgets.Button(
    description="Registrar vendedor",
    button_style="success",
    icon="check",
    layout=widgets.Layout(width="190px")
)


# Boton limpiar
boton_limpiar = widgets.Button(
    description="Limpiar",
    button_style="warning",
    icon="trash",
    layout=widgets.Layout(width="120px")
)


# Acciones de los botones
boton_registrar.on_click(registrar)
boton_limpiar.on_click(limpiar)


# Organizar los botones
botones = widgets.HBox(
    [boton_registrar, boton_limpiar],
    layout=widgets.Layout(
        justify_content="center",
        gap="15px"
    )
)


# Crear la interfaz
interfaz = widgets.VBox(
    [
        titulo,
        nombre,
        apellidos,
        edad,
        botones,
        salida
    ],
    layout=widgets.Layout(
        width="100%",
        padding="15px",
        gap="12px"
    )
)


# Mostrar interfaz
display(interfaz)
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output
import math


# Clase CalculosNumericos
class CalculosNumericos:

    # Metodo para calcular el logaritmo
    @staticmethod
    def calcular_logaritmo(valor):

        try:
            if valor < 0:
                raise ArithmeticError

            if valor == 0:
                return "-infinito"

            resultado = math.log(valor)
            return resultado

        except ArithmeticError:
            return "El valor debe ser positivo para calcular el logaritmo"

    # Metodo para calcular la raiz cuadrada
    @staticmethod
    def calcular_raiz(valor):

        try:
            if valor < 0:
                raise ArithmeticError

            resultado = math.sqrt(valor)
            return resultado

        except ArithmeticError:
            return "El valor debe ser positivo para calcular la raiz cuadrada"


# Titulo de la interfaz
titulo = widgets.HTML(
    value="""
    <h2 style="color:#FFFFFF; text-align:center;">
        Calculos numericos
    </h2>
    <p style="color:#FFFFFF; text-align:center;">
        Ejercicio 6.6. Catchs multiples
    </p>
    """
)


# Campo para ingresar el numero
numero = widgets.Text(
    description="Valor:",
    placeholder="Ingrese un numero",
    layout=widgets.Layout(width="90%")
)


# Espacio para mostrar los resultados
salida = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #75B798",
        padding="15px",
        width="95%"
    )
)


# Funcion para realizar los calculos
def calcular(b):

    with salida:
        clear_output()

        try:
            valor = float(numero.value)

            print("RESULTADOS")
            print()

            # Calcular logaritmo
            resultado_log = CalculosNumericos.calcular_logaritmo(valor)

            print("Logaritmo neperiano:")
            print(resultado_log)

            print()

            # Calcular raiz cuadrada
            resultado_raiz = CalculosNumericos.calcular_raiz(valor)

            print("Raiz cuadrada:")
            print(resultado_raiz)

        except ValueError:
            print("Error: El valor debe ser numerico para realizar los calculos")


# Funcion para limpiar
def limpiar(b):

    numero.value = ""

    with salida:
        clear_output()


# Boton calcular
boton_calcular = widgets.Button(
    description="Calcular",
    button_style="success",
    icon="calculator",
    layout=widgets.Layout(width="150px")
)


# Boton limpiar
boton_limpiar = widgets.Button(
    description="Limpiar",
    button_style="warning",
    icon="trash",
    layout=widgets.Layout(width="120px")
)


# Acciones de los botones
boton_calcular.on_click(calcular)
boton_limpiar.on_click(limpiar)


# Organizar los botones
botones = widgets.HBox(
    [boton_calcular, boton_limpiar],
    layout=widgets.Layout(
        justify_content="center",
        gap="15px"
    )
)


# Crear la interfaz
interfaz = widgets.VBox(
    [
        titulo,
        numero,
        botones,
        salida
    ],
    layout=widgets.Layout(
        width="100%",
        padding="15px",
        gap="12px"
    )
)


# Mostrar interfaz
display(interfaz)
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output


# Clase Programador
class Programador:

    def __init__(self, nombre, apellidos):
        self.nombre = nombre
        self.apellidos = apellidos


# Clase EquipoMaratonProgramacion
class EquipoMaratonProgramacion:

    def __init__(self, nombre, universidad, lenguaje):
        self.nombre = nombre
        self.universidad = universidad
        self.lenguaje = lenguaje
        self.programadores = []

    # Metodo para saber si el equipo esta lleno
    def esta_lleno(self):
        return len(self.programadores) == 3

    # Metodo para agregar un programador
    def anadir(self, programador):

        if self.esta_lleno():
            raise ValueError("El equipo esta completo. No se pueden agregar mas integrantes")

        self.programadores.append(programador)

    # Metodo para validar los campos
    @staticmethod
    def validar_campo(campo):

        if campo == "":
            raise ValueError("El campo no puede estar vacio")

        for caracter in campo:
            if caracter.isdigit():
                raise ValueError("El nombre no puede tener numeros")

        if len(campo) >= 20:
            raise ValueError("El nombre debe tener menos de 20 caracteres")


# Crear el equipo
equipo = None


# Titulo de la interfaz
titulo = widgets.HTML(
    value="""
    <h2 style="color:#FFFFFF; text-align:center;">
        Registro de equipo
    </h2>
    <p style="color:#FFFFFF; text-align:center;">
        Ejercicio 6.7. Validacion de campos
    </p>
    """
)


# Datos del equipo
nombre_equipo = widgets.Text(
    description="Equipo:",
    placeholder="Nombre del equipo",
    layout=widgets.Layout(width="90%")
)

universidad = widgets.Text(
    description="Universidad:",
    placeholder="Nombre de la universidad",
    layout=widgets.Layout(width="90%")
)

lenguaje = widgets.Text(
    description="Lenguaje:",
    placeholder="Lenguaje de programacion",
    layout=widgets.Layout(width="90%")
)


# Datos del programador
nombre_programador = widgets.Text(
    description="Nombre:",
    placeholder="Nombre del integrante",
    layout=widgets.Layout(width="90%")
)

apellidos_programador = widgets.Text(
    description="Apellidos:",
    placeholder="Apellidos del integrante",
    layout=widgets.Layout(width="90%")
)


# Espacio para mostrar resultados
salida = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #75B798",
        padding="15px",
        width="95%"
    )
)


# Funcion para crear el equipo
def crear_equipo(b):

    global equipo

    with salida:
        clear_output()

        try:
            n = nombre_equipo.value.strip()
            u = universidad.value.strip()
            l = lenguaje.value.strip()

            if n == "" or u == "" or l == "":
                raise ValueError("Debe completar todos los datos del equipo")

            equipo = EquipoMaratonProgramacion(n, u, l)

            print("Equipo creado correctamente")
            print("Nombre:", equipo.nombre)
            print("Universidad:", equipo.universidad)
            print("Lenguaje:", equipo.lenguaje)
            print("\nAhora puede agregar sus integrantes")

        except ValueError as error:
            print("Error:", error)


# Funcion para agregar integrantes
def agregar(b):

    with salida:
        clear_output()

        try:
            if equipo is None:
                raise ValueError("Primero debe crear el equipo")

            n = nombre_programador.value.strip()
            a = apellidos_programador.value.strip()

            # Validar nombre y apellidos
            EquipoMaratonProgramacion.validar_campo(n)
            EquipoMaratonProgramacion.validar_campo(a)

            # Crear el programador
            programador = Programador(n, a)

            # Agregar al equipo
            equipo.anadir(programador)

            print("Integrante agregado correctamente\n")
            print("INTEGRANTES DEL EQUIPO")

            for i, programador in enumerate(equipo.programadores):
                print(i + 1, "-", programador.nombre, programador.apellidos)

            print("\nTotal de integrantes:", len(equipo.programadores))

            nombre_programador.value = ""
            apellidos_programador.value = ""

        except ValueError as error:
            print("No se pudo agregar el integrante")
            print("Error:", error)


# Funcion para registrar el equipo
def registrar(b):

    with salida:
        clear_output()

        try:
            if equipo is None:
                raise ValueError("Primero debe crear el equipo")

            if len(equipo.programadores) < 2:
                raise ValueError("El equipo debe tener minimo 2 integrantes")

            print("EQUIPO REGISTRADO CORRECTAMENTE\n")

            print("Nombre del equipo:", equipo.nombre)
            print("Universidad:", equipo.universidad)
            print("Lenguaje:", equipo.lenguaje)

            print("\nINTEGRANTES:")

            for i, programador in enumerate(equipo.programadores):
                print(i + 1, "-", programador.nombre, programador.apellidos)

            print("\nTotal de integrantes:", len(equipo.programadores))

        except ValueError as error:
            print("No se pudo registrar el equipo")
            print("Error:", error)


# Funcion para limpiar
def limpiar(b):

    global equipo
    equipo = None

    nombre_equipo.value = ""
    universidad.value = ""
    lenguaje.value = ""
    nombre_programador.value = ""
    apellidos_programador.value = ""

    with salida:
        clear_output()


# Botones
boton_crear = widgets.Button(
    description="Crear equipo",
    button_style="success",
    icon="check",
    layout=widgets.Layout(width="150px")
)

boton_agregar = widgets.Button(
    description="Agregar integrante",
    button_style="success",
    icon="user-plus",
    layout=widgets.Layout(width="180px")
)

boton_registrar = widgets.Button(
    description="Registrar equipo",
    button_style="success",
    icon="save",
    layout=widgets.Layout(width="160px")
)

boton_limpiar = widgets.Button(
    description="Limpiar",
    button_style="warning",
    icon="trash",
    layout=widgets.Layout(width="120px")
)


# Acciones de los botones
boton_crear.on_click(crear_equipo)
boton_agregar.on_click(agregar)
boton_registrar.on_click(registrar)
boton_limpiar.on_click(limpiar)


# Organizar los botones
botones_equipo = widgets.HBox(
    [boton_crear],
    layout=widgets.Layout(
        justify_content="center"
    )
)

botones_integrantes = widgets.HBox(
    [boton_agregar, boton_registrar, boton_limpiar],
    layout=widgets.Layout(
        justify_content="center",
        gap="10px",
        flex_flow="row wrap"
    )
)


# Crear la interfaz
interfaz = widgets.VBox(
    [
        titulo,

        widgets.HTML(
            "<h3 style='color:#FFFFFF;'>Datos del equipo</h3>"
        ),

        nombre_equipo,
        universidad,
        lenguaje,
        botones_equipo,

        widgets.HTML(
            "<h3 style='color:#FFFFFF;'>Datos de los integrantes</h3>"
        ),

        nombre_programador,
        apellidos_programador,
        botones_integrantes,

        salida
    ],
    layout=widgets.Layout(
        width="100%",
        padding="15px",
        gap="12px"
    )
)


# Mostrar interfaz
display(interfaz)
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output


class LeerArchivo:

    @staticmethod
    def leer_archivo(nombre):
        try:
            archivo = open(nombre, "r", encoding="utf-8")

            linea = archivo.readline()

            while linea != "":
                print(linea, end="")
                linea = archivo.readline()

            archivo.close()

        except FileNotFoundError:
            print("No se encontró el archivo.")

        except:
            print("No se pudo leer el archivo.")


# Titulo
titulo = widgets.HTML(
    "<h2 style='color:white;'>Lectura de archivos</h2>"
)

descripcion = widgets.HTML(
    "<p style='color:white;'>Escribe el nombre del archivo que deseas leer.</p>"
)

# Campo para escribir el nombre
nombre = widgets.Text(
    placeholder="Ejemplo: Prueba 1.txt",
    description="Archivo:",
    layout=widgets.Layout(width="450px")
)

# Botones
boton_leer = widgets.Button(
    description="Leer archivo",
    button_style="success",
    layout=widgets.Layout(width="170px")
)

boton_limpiar = widgets.Button(
    description="Limpiar",
    button_style="warning",
    layout=widgets.Layout(width="170px")
)

# Espacio para mostrar resultados
salida = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #888",
        padding="12px",
        width="550px"
    )
)


def leer(b):
    with salida:
        clear_output()

        nombre_archivo = nombre.value.strip()

        if nombre_archivo == "":
            print("Escribe el nombre del archivo.")
        else:
            print("CONTENIDO DEL ARCHIVO")
            print()

            ruta = "/content/sample_data/" + nombre_archivo

            LeerArchivo.leer_archivo(ruta)


def limpiar(b):
    nombre.value = ""

    with salida:
        clear_output()


boton_leer.on_click(leer)
boton_limpiar.on_click(limpiar)


# Mostrar interfaz
display(
    widgets.VBox(
        [
            titulo,
            descripcion,
            nombre,
            widgets.HBox([boton_leer, boton_limpiar]),
            salida
        ],
        layout=widgets.Layout(
            background_color="#263238",
            padding="20px",
            width="600px"
        )
    )
)
```
